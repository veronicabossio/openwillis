```python
%reload_ext autoreload
%autoreload 2

from lexical import (
    get_mattr,
    get_brunet_index,   
    get_honore_statistic,
    score_excerpt,
    get_parented_tree
)

import time
import spacy
from nltk.tree import ParentedTree
from spacy.tokens import Span
```

```python
from pathlib import Path

nlp1 = spacy.load("en_core_web_sm")
#nlp = spacy.load("en_core_web_sm")

# if "benepar" not in nlp.pipe_names:
#     nlp.add_pipe("benepar", config={"model": "benepar_en3"})

if "benepar" not in nlp1.pipe_names:
    nlp1.add_pipe("benepar", config={"model": "benepar_en3"})

if not Span.has_extension("parse_tree"):
    Span.set_extension("parse_tree", getter=get_parented_tree)

results = []

for file_path in Path("/Users/veronicabossio/openwillis/openwillis-speech/validation_notebooks").glob("*.txt"):
    with open(file_path, "r") as f:
        text = f.read()

    start = time.time()
    _ = get_mattr(text, nlp1)
    _ = get_brunet_index(text, nlp1)
    _ = get_honore_statistic(text, nlp1)
    lexical_time = time.time() - start

    start = time.time()
    _ = get_mattr(text, nlp1)
    _ = get_brunet_index(text, nlp1)
    _ = get_honore_statistic(text, nlp1)
    _ = score_excerpt(text, nlp1, nlp1)
    full_time = time.time() - start

    ratio = full_time / lexical_time if lexical_time > 0 else float('inf')
    results.append((file_path.name, lexical_time, full_time, ratio))

for r in results:
    print(f"{r[0]}: lexical={r[1]:.3f}s, full={r[2]:.3f}s, ratio={r[3]:.2f}x")
```
